# 1. Introduction to LLM APIs

A Large Language Model (LLM) API allows applications to communicate with an AI language model programmatically.

Instead of interacting with an AI model only through a website, developers can send requests from their own applications and receive AI-generated responses.

LLM APIs can be used to build:

- AI chatbots
- Virtual assistants
- Content generation tools
- Summarization applications
- Question-answering systems
- Coding assistants
- Educational applications

---

# 2. How an LLM API Works

A basic API workflow is:

**Application → API Request → LLM → API Response → Application**

The application sends information to the API.

The API sends the request to the selected language model.

The model processes the request and generates a response.

The response is then returned to the application.

---

# 3. Important API Concepts

## API Key

An API key is a credential used to authenticate an application with an API service.

API keys should be treated as sensitive information.

Never publish an API key in:

- GitHub repositories
- Screenshots
- README files
- Public notebooks
- Social media posts

---

## Authentication

Authentication verifies that an application is authorized to access an API.

A common approach is:

**Application → API Key → API Service → Authentication → Response**

---

## Environment Variables

Environment variables provide a safer way to store sensitive configuration values such as API keys.

Instead of writing a key directly inside source code, applications can retrieve it from an environment variable.

Example:

`API_KEY`

The actual secret value should remain private.

---

# 4. Why API Keys Should Be Protected

Exposing an API key can allow unauthorized users to use the associated API account.

Possible consequences include:

- Unauthorized API usage
- Unexpected charges
- Account security problems
- Abuse of API resources

Therefore, API keys should never be hard-coded into publicly shared projects.

---

# 5. LLM API Request Structure

A typical LLM request may contain:

- Model name
- User message
- System instructions
- Conversation history
- Generation parameters

A simplified workflow is:

**Input → API Request → Model Processing → Generated Response**

---

# 6. System, User and Assistant Messages

Example:

"You are a helpful programming tutor."

## User Message

Contains the user's request.

## Assistant Message

Example:

# 7. Model Parameters

LLM APIs may provide parameters that influence generation.

Important parameters can include:

### Temperature

Controls the randomness of generated responses.

Lower values generally produce more predictable responses.

Higher values can produce more varied responses.

### Maximum Output Tokens

Controls the maximum amount of generated output.

### Model

Specifies which language model should process the request.

The available parameters depend on the API provider and model.

# 12. Introduction to Conversation Management

Conversation management is the process of maintaining and organizing messages exchanged between a user and an AI model.

A conversational AI application usually needs to remember previous messages so that the model can understand the context of the conversation.

A typical conversation contains:

- System messages
- User messages
- Assistant messages

The conversation can be represented as a sequence of messages.

---

# 13. Message Roles

## System Message

A system message defines the behavior, role or instructions for the AI.

Example:

"You are a helpful AI tutor."

## User Message

A user message contains the request or question from the user.

Example:

"Explain machine learning."

## Assistant Message

An assistant message contains the response generated by the AI.

Example:

"Machine learning is a method that allows computers to learn patterns from data."

---

# 14. Chat History

Chat history stores previous messages exchanged during a conversation.

Example:

User:
"Explain Python."

Assistant:
"Python is a programming language..."

User:
"What is a variable?"

The previous conversation helps the AI understand that the user is still asking about Python.

---

# 15. Why Chat History Is Important

Chat history helps applications provide:

- Context-aware responses
- Follow-up question support
- Continuous conversations
- Personalized interactions
- Better understanding of previous messages

Without conversation history, each request may be treated as an independent question.

---

# 16. Managing Conversation History in Python

            "content": "What is Python?"
        },
        {
            "role": "assistant",
            "content": "Python is a programming language."
        }

New messages can be added to the list as the conversation continues.

This process is repeated for every new user message.

```python
# Conversation Management Experiment

messages = [
    {
        "role": "system",
        "content": "You are a helpful AI study assistant. Explain concepts clearly for a B.Tech student."
    },
    {
        "role": "user",
        "content": "What is machine learning?"
    }
]

response = client.chat.completions.create(
    model="openai/gpt-oss-120b:fastest",
    messages=messages,
    max_tokens=150
)

assistant_reply = response.choices[0].message.content

print("Assistant:", assistant_reply)

# Add the assistant response to chat history
messages.append({
    "role": "assistant",
    "content": assistant_reply
})

print("\n--- Chat History ---")
for message in messages:
    print(message["role"].upper() + ":", message["content"])
```

```python
# Follow-up Question Using Chat History

messages.append({
    "role": "user",
    "content": "Can you give me a simple real-world example?"
})

response = client.chat.completions.create(
    model="openai/gpt-oss-120b:fastest",
    messages=messages,
    max_tokens=150
)

assistant_reply = response.choices[0].message.content

print("Assistant:", assistant_reply)

# Add the new assistant response to the conversation history
messages.append({
    "role": "assistant",
    "content": assistant_reply
})

print("\n--- Updated Chat History ---")
for message in messages:
    print(message["role"].upper() + ":", message["content"])
```

```python
# Model Parameters and Token Usage

response = client.chat.completions.create(
    model="openai/gpt-oss-120b:fastest",
    messages=[
        {
            "role": "system",
            "content": "You are a concise AI tutor."
        },
        {
            "role": "user",
            "content": "Explain neural networks in simple words."
        }
    ],
    max_tokens=100
)

print("AI Response:")
print(response.choices[0].message.content)

print("\n--- Usage Information ---")

if response.usage:
    print("Input tokens:", response.usage.prompt_tokens)
    print("Output tokens:", response.usage.completion_tokens)
    print("Total tokens:", response.usage.total_tokens)
else:
    print("Token usage information was not returned by the provider.")
```

```python
# AI Chat Application - Colab Prototype

import os
from huggingface_hub import InferenceClient

client = InferenceClient(
    api_key=os.environ["HF_TOKEN"]
)

SYSTEM_INSTRUCTION = """
You are a helpful AI study assistant for a B.Tech AIML student.
Explain technical concepts clearly using simple language and examples.
"""

chat_history = [
    {
        "role": "system",
        "content": SYSTEM_INSTRUCTION
    }
]

def chat_with_ai(user_message):
    chat_history.append({
        "role": "user",
        "content": user_message
    })

    try:
        response = client.chat.completions.create(
            model="openai/gpt-oss-120b:fastest",
            messages=chat_history,
            max_tokens=300
        )

        assistant_reply = response.choices[0].message.content

        if not assistant_reply:
            assistant_reply = "The AI did not return a response."

        chat_history.append({
            "role": "assistant",
            "content": assistant_reply
        })

        return assistant_reply

    except Exception as error:
        return f"Sorry, an error occurred: {type(error).__name__}"


print("AI Chat Application Ready!")
print("Type a question when prompted.")

user_question = input("You: ")

answer = chat_with_ai(user_question)

print("\nAI:", answer)
```

```python
# Clear Chat History

def clear_chat():
    global chat_history

    chat_history = [
        {
            "role": "system",
            "content": SYSTEM_INSTRUCTION
        }
    ]

    print("Chat history cleared successfully.")


# Test the clear-chat feature
clear_chat()

print("\nCurrent Chat History:")
print(chat_history)
```

```python
# Multi-Turn Conversation Test

clear_chat()

questions = [
    "What is machine learning?",
    "Give me one real-world example.",
    "What is the main benefit of that example?"
]

for question in questions:
    print("\nYou:", question)

    answer = chat_with_ai(question)

    print("AI:", answer)

print("\n--- Final Chat History ---")

for message in chat_history:
    print(message["role"].upper() + ":", message["content"])
```
